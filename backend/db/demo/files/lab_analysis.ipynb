# 応答時間の分析（デモ用）
条件ごとに、資料を探し出すまでの時間を比べる。

```
import csv, statistics
rows = list(csv.DictReader(open('lab_response_time.csv', encoding='utf-8')))
len(rows)
```

```
for cond in ['フォルダ整理あり', 'フォルダ整理なし', 'タグ検索']:
    xs = [int(r['応答時間(ms)']) for r in rows if r['条件'] == cond]
    print(cond, round(statistics.mean(xs)))
```

**結論**：整理とタグ検索で、探す時間はおよそ半分になった。
